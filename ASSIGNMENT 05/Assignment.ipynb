### 1.3.2 Gender

```python
print('Gender Value Counts:')
print(loan_df['Gender'].value_counts(dropna=False))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
loan_df['Gender'].value_counts().plot(kind='bar', ax=axes[0], color=['#4B8BBE', '#FF6B6B'])
axes[0].set_title('Gender Distribution')
axes[0].set_xlabel('Gender')
axes[0].set_ylabel('Count')
for i, v in enumerate(loan_df['Gender'].value_counts().values):
    axes[0].text(i, v + 5, str(v), ha='center')
loan_df['Gender'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%', colors=['#4B8BBE', '#FF6B6B'], startangle=90)
axes[1].set_title('Gender Proportion')
axes[1].set_ylabel('')
plt.tight_layout()
plt.show()

print(f'Missing Gender: {loan_df["Gender"].isnull().sum()}')
loan_df['Gender'].fillna(loan_df['Gender'].mode()[0], inplace=True)
print(f'After filling: {loan_df["Gender"].isnull().sum()} missing')
```

```python
print('Married Value Counts:')
print(loan_df['Married'].value_counts(dropna=False))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
loan_df['Married'].value_counts().plot(kind='bar', ax=axes[0], color=['#2ECC71', '#E74C3C', '#95A5A6'])
axes[0].set_title('Married Distribution')
axes[0].set_xlabel('Married')
axes[0].set_ylabel('Count')
for i, v in enumerate(loan_df['Married'].value_counts().values):
    axes[0].text(i, v + 5, str(v), ha='center')
loan_df['Married'].value_counts().plot(kind='pie', ax=axes[1], autopct='%1.1f%%', colors=['#2ECC71', '#E74C3C', '#95A5A6'], startangle=90)
axes[1].set_title('Married Proportion')
axes[1].set_ylabel('')
plt.tight_layout()
plt.show()

print(f'Missing Married: {loan_df["Married"].isnull().sum()}')
loan_df['Married'].fillna(loan_df['Married'].mode()[0], inplace=True)
print(f'After filling: {loan_df["Married"].isnull().sum()} missing')
```

### 1.3.4 Dependents

```python
print('Dependents Value Counts:')
print(loan_df['Dependents'].value_counts(dropna=False))

fig, ax = plt.subplots(figsize=(10, 5))
loan_df['Dependents'].value_counts().plot(kind='bar', ax=ax, color='#9B59B6')
ax.set_title('Dependents Distribution')
ax.set_xlabel('Dependents')
ax.set_ylabel('Count')
for i, v in enumerate(loan_df['Dependents'].value_counts().values):
    ax.text(i, v + 5, str(v), ha='center')
plt.show()

loan_df['Dependents'] = loan_df['Dependents'].replace('3+', '3').astype(str)
print(f'Missing Dependents: {loan_df["Dependents"].isnull().sum()}')
loan_df['Dependents'].fillna(loan_df['Dependents'].mode()[0], inplace=True)
print(f'After filling: {loan_df["Dependents"].isnull().sum()} missing')
```

### 1.3.7 ApplicantIncome

```python
print('ApplicantIncome Stats:')
print(loan_df['ApplicantIncome'].describe())

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
axes[0].hist(loan_df['ApplicantIncome'], bins=50, color='#3498DB', edgecolor='black', alpha=0.7)
axes[0].set_title('Applicant Income Distribution')
axes[0].set_xlabel('Income')
axes[0].set_ylabel('Frequency')
axes[1].boxplot(loan_df['ApplicantIncome'])
axes[1].set_title('Applicant Income Box Plot')
axes[1].set_ylabel('Income')
plt.tight_layout()
plt.show()

Q1 = loan_df['ApplicantIncome'].quantile(0.25)
Q3 = loan_df['ApplicantIncome'].quantile(0.75)
IQR = Q3 - Q1
lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR
outliers = loan_df[(loan_df['ApplicantIncome'] < lower) | (loan_df['ApplicantIncome'] > upper)]
print(f'Outliers (IQR method): {len(outliers)} rows ({len(outliers)/len(loan_df)*100:.2f}%)')
print(f'Missing ApplicantIncome: {loan_df["ApplicantIncome"].isnull().sum()}')
```

## 1.4 Bivariate Analysis

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 12))
cats = ['Gender', 'Married', 'Education', 'Self_Employed', 'Credit_History', 'Property_Area']
for i, col in enumerate(cats):
    ax = axes[i//3][i%3]
    pd.crosstab(loan_df[col], loan_df['Loan_Status']).plot(kind='bar', ax=ax, color=['#E74C3C', '#27AE60'])
    ax.set_title(f'Loan Status by {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Count')
    ax.legend(['N', 'Y'])
plt.tight_layout()
plt.show()
```

## 1.5 Correlation Analysis

```python
loan_enc = loan_df.copy()
loan_enc['Gender'] = loan_enc['Gender'].map({'Male': 1, 'Female': 0})
loan_enc['Married'] = loan_enc['Married'].map({'Yes': 1, 'No': 0})
loan_enc['Education'] = loan_enc['Education'].map({'Graduate': 1, 'Not Graduate': 0})
loan_enc['Self_Employed'] = loan_enc['Self_Employed'].map({'Yes': 1, 'No': 0})
loan_enc['Property_Area'] = loan_enc['Property_Area'].map({'Urban': 2, 'Semiurban': 1, 'Rural': 0})
loan_enc['Loan_Status'] = loan_enc['Loan_Status'].map({'Y': 1, 'N': 0})
loan_enc['Dependents'] = pd.to_numeric(loan_enc['Dependents'], errors='coerce')

plt.figure(figsize=(12, 8))
corr = loan_enc.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Loan Dataset - Correlation Heatmap')
plt.tight_layout()
plt.show()
print('Top correlations with Loan_Status:')
print(corr['Loan_Status'].drop('Loan_Status').sort_values(ascending=False))
```

```python
print('Cancellation by Hotel Type:')
print(hotel_df.groupby('hotel')['is_canceled'].mean()*100)
print()
print('Cancellation by Deposit Type:')
print(hotel_df.groupby('deposit_type')['is_canceled'].mean()*100)

fig, axes = plt.subplots(2, 3, figsize=(18, 12))
cats = ['hotel', 'meal', 'market_segment', 'deposit_type', 'customer_type', 'reservation_status']
for i, col in enumerate(cats):
    ax = axes[i//3][i%3]
    pd.crosstab(hotel_df[col], hotel_df['is_canceled']).plot(kind='bar', ax=ax, color=['#2ECC71', '#E74C3C'], stacked=True)
    ax.set_title(f'Cancellation by {col}')
    ax.set_xlabel(col)
    ax.set_ylabel('Count')
    ax.legend(['Not Canceled', 'Canceled'])
    ax.tick_params(axis='x', rotation=45)
plt.tight_layout()
plt.show()
```

## 3.5 Correlation Analysis

```python
hotel_enc = hotel_df.copy()
hotel_enc['hotel'] = hotel_enc['hotel'].map({'Resort Hotel': 1, 'City Hotel': 0})
hotel_enc['meal'] = hotel_enc['meal'].map({'BB': 1, 'HB': 2, 'FB': 3, 'SC': 0, 'Undefined': -1})
hotel_enc['deposit_type'] = hotel_enc['deposit_type'].map({'No Deposit': 0, 'Non Refund': 1, 'Refundable': 2})
hotel_enc['customer_type'] = hotel_enc['customer_type'].map({'Transient': 0, 'Transient-Party': 1, 'Contract': 2, 'Group': 3})

plt.figure(figsize=(16, 12))
corr = hotel_enc.select_dtypes(include=[np.number]).corr()
sns.heatmap(corr, annot=True, cmap='coolwarm', fmt='.2f', linewidths=0.5)
plt.title('Hotel Booking - Correlation Heatmap')
plt.tight_layout()
plt.show()
print('Top correlations with Is_Canceled:')
print(corr['is_canceled'].drop('is_canceled').sort_values(ascending=False).head(10))
```
