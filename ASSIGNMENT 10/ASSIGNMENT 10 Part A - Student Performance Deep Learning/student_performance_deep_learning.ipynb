# Assignment 10A: Student Performance Prediction Using Deep Learning

This notebook builds ANN models to predict student final grades using various activation functions and optimizers.

```python
import warnings
warnings.filterwarnings('ignore')
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

print('TensorFlow version:', tf.__version__)
```

```python
df = pd.read_csv('student_performance_prediction_dataset-2.csv')
```

```python
df.head()
```

```python
df.info()
print(df.shape)
```

```python
print("Missing Values:")
print(df.isnull().sum())
print("\nDuplicate Rows:")
print(df.duplicated().sum())
```

```python
categorical_cols = df.select_dtypes(include=['object']).columns
print("Categorical Columns:", categorical_cols.tolist())
for col in categorical_cols:
    print(f"\n{col}:")
    print(df[col].value_counts())
```

```python
df_encoded = df.copy()
for col in categorical_cols:
    df_encoded[col] = pd.Categorical(df_encoded[col]).codes
```

```python
X = df_encoded.drop('final_grade', axis=1)
y = df_encoded['final_grade']
print("Features shape:", X.shape)
print("Target shape:", y.shape)
```

```python
X_sample = X.sample(n=50000, random_state=42)
y_sample = y[X_sample.index]

X_train, X_test, y_train, y_test = train_test_split(X_sample, y_sample, test_size=0.2, random_state=42)
print(f"Training set size: {X_train.shape[0]}")
print(f"Testing set size: {X_test.shape[0]}")
```

```python
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

## Experiment 1: ReLU + Adam Optimizer

```python
model_a = keras.Sequential([
    layers.Dense(128, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    layers.Dense(64, activation='relu'),
    layers.Dense(32, activation='relu'),
    layers.Dense(1)
])
model_a.compile(optimizer='adam', loss='mse', metrics=['mae'])
history_a = model_a.fit(X_train_scaled, y_train, epochs=15, batch_size=64, validation_split=0.2, verbose=0)
y_pred_a = model_a.predict(X_test_scaled, verbose=0)
mse_a = mean_squared_error(y_test, y_pred_a)
rmse_a = np.sqrt(mse_a)
mae_a = mean_absolute_error(y_test, y_pred_a)
r2_a = r2_score(y_test, y_pred_a)
print(f"Model A (ReLU-Adam): MSE={mse_a:.4f}, RMSE={rmse_a:.4f}, MAE={mae_a:.4f}, R2={r2_a:.4f}")
```

## Experiment 2: Tanh + SGD Optimizer

```python
model_b = keras.Sequential([
    layers.Dense(100, activation='tanh', input_shape=(X_train_scaled.shape[1],)),
    layers.Dense(50, activation='tanh'),
    layers.Dense(25, activation='tanh'),
    layers.Dense(1)
])
model_b.compile(optimizer='sgd', loss='mse', metrics=['mae'])
history_b = model_b.fit(X_train_scaled, y_train, epochs=15, batch_size=64, validation_split=0.2, verbose=0)
y_pred_b = model_b.predict(X_test_scaled, verbose=0)
mse_b = mean_squared_error(y_test, y_pred_b)
rmse_b = np.sqrt(mse_b)
mae_b = mean_absolute_error(y_test, y_pred_b)
r2_b = r2_score(y_test, y_pred_b)
print(f"Model B (Tanh-SGD): MSE={mse_b:.4f}, RMSE={rmse_b:.4f}, MAE={mae_b:.4f}, R2={r2_b:.4f}")
```

## Experiment 3: ReLU + RMSprop Optimizer

```python
model_c = keras.Sequential([
    layers.Dense(120, activation='relu', input_shape=(X_train_scaled.shape[1],)),
    layers.Dense(60, activation='relu'),
    layers.Dense(30, activation='relu'),
    layers.Dense(1)
])
model_c.compile(optimizer='rmsprop', loss='mse', metrics=['mae'])
history_c = model_c.fit(X_train_scaled, y_train, epochs=15, batch_size=64, validation_split=0.2, verbose=0)
y_pred_c = model_c.predict(X_test_scaled, verbose=0)
mse_c = mean_squared_error(y_test, y_pred_c)
rmse_c = np.sqrt(mse_c)
mae_c = mean_absolute_error(y_test, y_pred_c)
r2_c = r2_score(y_test, y_pred_c)
print(f"Model C (ReLU-RMSprop): MSE={mse_c:.4f}, RMSE={rmse_c:.4f}, MAE={mae_c:.4f}, R2={r2_c:.4f}")
```

## Experiment 4: Sigmoid + Adam Optimizer

```python
model_d = keras.Sequential([
    layers.Dense(96, activation='sigmoid', input_shape=(X_train_scaled.shape[1],)),
    layers.Dense(48, activation='sigmoid'),
    layers.Dense(24, activation='sigmoid'),
    layers.Dense(1)
])
model_d.compile(optimizer='adam', loss='mse', metrics=['mae'])
history_d = model_d.fit(X_train_scaled, y_train, epochs=15, batch_size=64, validation_split=0.2, verbose=0)
y_pred_d = model_d.predict(X_test_scaled, verbose=0)
mse_d = mean_squared_error(y_test, y_pred_d)
rmse_d = np.sqrt(mse_d)
mae_d = mean_absolute_error(y_test, y_pred_d)
r2_d = r2_score(y_test, y_pred_d)
print(f"Model D (Sigmoid-Adam): MSE={mse_d:.4f}, RMSE={rmse_d:.4f}, MAE={mae_d:.4f}, R2={r2_d:.4f}")
```

## Experiment 5: Tanh + RMSprop Optimizer

```python
model_e = keras.Sequential([
    layers.Dense(110, activation='tanh', input_shape=(X_train_scaled.shape[1],)),
    layers.Dense(55, activation='tanh'),
    layers.Dense(27, activation='tanh'),
    layers.Dense(1)
])
model_e.compile(optimizer='rmsprop', loss='mse', metrics=['mae'])
history_e = model_e.fit(X_train_scaled, y_train, epochs=15, batch_size=64, validation_split=0.2, verbose=0)
y_pred_e = model_e.predict(X_test_scaled, verbose=0)
mse_e = mean_squared_error(y_test, y_pred_e)
rmse_e = np.sqrt(mse_e)
mae_e = mean_absolute_error(y_test, y_pred_e)
r2_e = r2_score(y_test, y_pred_e)
print(f"Model E (Tanh-RMSprop): MSE={mse_e:.4f}, RMSE={rmse_e:.4f}, MAE={mae_e:.4f}, R2={r2_e:.4f}")
```

## Experiment 6: Sigmoid + SGD Optimizer

```python
model_f = keras.Sequential([
    layers.Dense(105, activation='sigmoid', input_shape=(X_train_scaled.shape[1],)),
    layers.Dense(52, activation='sigmoid'),
    layers.Dense(26, activation='sigmoid'),
    layers.Dense(1)
])
model_f.compile(optimizer='sgd', loss='mse', metrics=['mae'])
history_f = model_f.fit(X_train_scaled, y_train, epochs=15, batch_size=64, validation_split=0.2, verbose=0)
y_pred_f = model_f.predict(X_test_scaled, verbose=0)
mse_f = mean_squared_error(y_test, y_pred_f)
rmse_f = np.sqrt(mse_f)
mae_f = mean_absolute_error(y_test, y_pred_f)
r2_f = r2_score(y_test, y_pred_f)
print(f"Model F (Sigmoid-SGD): MSE={mse_f:.4f}, RMSE={rmse_f:.4f}, MAE={mae_f:.4f}, R2={r2_f:.4f}")
```

## Model Comparison

```python
comparison_data = {
    'Model': ['Model A', 'Model B', 'Model C', 'Model D', 'Model E', 'Model F'],
    'Activation': ['ReLU', 'Tanh', 'ReLU', 'Sigmoid', 'Tanh', 'Sigmoid'],
    'Optimizer': ['Adam', 'SGD', 'RMSprop', 'Adam', 'RMSprop', 'SGD'],
    'Epochs': [15, 15, 15, 15, 15, 15],
    'MSE': [mse_a, mse_b, mse_c, mse_d, mse_e, mse_f],
    'RMSE': [rmse_a, rmse_b, rmse_c, rmse_d, rmse_e, rmse_f],
    'MAE': [mae_a, mae_b, mae_c, mae_d, mae_e, mae_f],
    'R2 Score': [r2_a, r2_b, r2_c, r2_d, r2_e, r2_f]
}
comparison_df = pd.DataFrame(comparison_data)
print(comparison_df.to_string(index=False))
```

## Training History Visualization

```python
fig, axes = plt.subplots(2, 3, figsize=(18, 10))
histories = [history_a, history_b, history_c, history_d, history_e, history_f]
titles = ['Model A: ReLU-Adam', 'Model B: Tanh-SGD', 'Model C: ReLU-RMSprop',
          'Model D: Sigmoid-Adam', 'Model E: Tanh-RMSprop', 'Model F: Sigmoid-SGD']

for ax, hist, title in zip(axes.flat, histories, titles):
    ax.plot(hist.history['loss'], label='Training Loss')
    ax.plot(hist.history['val_loss'], label='Validation Loss')
    ax.set_title(title)
    ax.set_xlabel('Epoch')
    ax.set_ylabel('Loss')
    ax.legend()

plt.tight_layout()
plt.show()
```

## Final Predictions with Best Model

```python
best_idx = comparison_df['R2 Score'].idxmax()
best_model_name = comparison_df.loc[best_idx, 'Model']
print(f"Best Model: {best_model_name}")

all_preds = [y_pred_a, y_pred_b, y_pred_c, y_pred_d, y_pred_e, y_pred_f]
best_predictions = all_preds[best_idx]

print("\nSample Predictions (Actual vs Predicted):")
sample_results = pd.DataFrame({
    'Actual': y_test.values[:10],
    'Predicted': best_predictions.flatten()[:10]
})
print(sample_results.to_string(index=False))
```
